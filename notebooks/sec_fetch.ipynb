```python
from edgar import Company, set_identity

set_identity("surajonlyforgames@gmail.com")

company = Company("AAPL")
filings = company.get_filings(
    form="10-K",
    filing_date="2023-01-01:"
)
count = 1
for filing in filings:

    # Basic information
    print("Form: ", count)
    count += 1
    print(f"Form type: {filing.form}")
    print(f"Filing date: {filing.filing_date}")
    print(f"Accession number: {filing.accession_no}")
    print(f"Company: {filing.company}")
    print(f"CIK: {filing.cik}")
    # URLs
    print(f"Filing URL: {filing.homepage_url}")
    print(f"Document URL: {filing.document.url}")
    # Period information
    print(f"Period of report: {filing.period_of_report}")
```
